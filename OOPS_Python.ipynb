Q)Create a class circle with a constructur initializing a radus,another function area to calculate the area of the circle, another function to calculate the circumference of the circle

Q)there are two classes namely num1 and num2 .in num1 class there are 2 variables a and b that needs to be initialised using constructor also have a member function show value to print this value or display this values.num2 class is inherited fron num 1 and it has two functions 1st is show some that display the sum of the variables ,the 2nd one is show product that display the product of the variable

```python
class num1:
  def __init__(self,a,b):
    self.a=a
    self.b=b
  def show_value(self):
    print(f"value of a is {self.a}")
    print(f"value of a is {self.b}")
class num2(num1):
  def show_sum(self):
    print(self.a+self.b)
  def show_product(self):
    print(self.a*self.b)
n1=num2(2,6)
n1.show_value()
n1.show_sum()
n1.show_product()
```

```python
class student:
  def __init__(self,id,mark):
    self.id=id
    self.mark=mark
  def show_result(self):
    if self.mark>=40:
      print(f"student {self.id} passed")
    else:
      print(f"student {self.id} failed")
class master(student):
  def show_result1(self):
    if self.mark>=50:
      print(f"student {self.id} passed")
    else:
      print(f"student {self.id} failed")
s1=master(120,49)
s1.show_result()
m1=master(110,69)
m1.show_result1()
```
